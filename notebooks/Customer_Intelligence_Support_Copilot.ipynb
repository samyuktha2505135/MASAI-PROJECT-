# Instamart Customer Intelligence & Support Copilot
This exploratory notebook reads data prepared by `python run_project.py`. All source pipelines are reproducible `.py` files.

```python
from pathlib import Path
import pandas as pd
ROOT=Path.cwd().parent if Path.cwd().name=="notebooks" else Path.cwd()
orders=pd.read_csv(ROOT/"data/clean/orders.csv")
customers=pd.read_csv(ROOT/"data/clean/customers.csv")
print("Customers:",len(customers),"Orders:",len(orders))
```

## Part A: data validation and exploratory analysis

```python
print(orders.status.value_counts())
print("Completed revenue (INR):",orders.loc[orders.status.eq("completed"),"order_amount"].sum())
```

## Part B: customer segmentation

```python
segments=pd.read_csv(ROOT/"outputs/customer_segments.csv")
print(segments.groupby("segment_name").agg(customers=("customer_id","count"),mean_spend=("monetary","mean")))
```

## Part C: support triage

```python
triage=pd.read_csv(ROOT/"outputs/triage_results.csv")
print(triage.status.value_counts())
```
